## 3. Limpieza rigurosa

En esta sección se aplicarán transformaciones para resolver los problemas detectados durante el diagnóstico inicial, incluyendo valores faltantes, registros duplicados y tipos de datos que requieran corrección.

```python
print(df.duplicated().sum())
```

### Duplicados

Se verificó la existencia de registros completamente duplicados mediante `df.duplicated().sum()`.

El resultado fue 0, por lo que el dataset no presenta registros duplicados completos y no es necesario aplicar `drop_duplicates()`.

```python
print(df[df["provincia"].isna()])
```

### Tratamiento de valores faltantes

El diagnóstico inicial identificó 8 registros con valores faltantes en las variables
provincia, provincia_id, departamento y departamento_id.

Dado que estos registros representan aproximadamente el 0,005 % del dataset,
su eliminación no afecta de manera significativa el volumen de información disponible.

No se imputaron valores, ya que no existe información suficiente para determinar
de manera confiable qué provincia o departamento corresponde a estos registros.

Luego de inspeccionar los registros afectados, se decidió eliminarlos del dataset
utilizado para el análisis.

```python
df = df.dropna()
print(df.isna().sum())
```

```python
df = df.reset_index(drop=True)
print(df.index)
```

```python
print(df.dtypes)
# Comprobacion de datos nuevamente para su tranformacion correspondiente 
```

### Transformacion de datos 


```python
df["provincia_id"] = df["provincia_id"].astype(int)
df["departamento_id"] = df["departamento_id"].astype(int)
print(df.dtypes)
```

```python
print(df["provincia"].unique())
# Controlamos que no exista alguna existencia de algun dato repetido por una mala escritura o formato del string 
```

## 4. Agregaciones de negocio

En esta etapa se realizan agrupaciones para obtener información resumida del comportamiento de la producción agrícola.

Se analizan tres aspectos principales: la producción total por cultivo, la producción total por provincia y el rendimiento promedio por cultivo.

Estas agrupaciones permiten observar el volumen de producción y el rendimiento desde diferentes dimensiones del dataset, facilitando una primera interpretación de los datos.

```python
# ¿Cuanta produccion total se obtiene por cada cultivo?
print(df.groupby("cultivo")["produccion_tm"].sum())
```

```python
# ¿Que cantidad de produccion total registra cada provincia?
print(df.groupby("provincia")["produccion_tm"].sum())
```

```python
# ¿Cual es el rendimiento promedio de cada cultivo?
print(df.groupby("cultivo")["rendimiento_kgxha"].mean())
```
